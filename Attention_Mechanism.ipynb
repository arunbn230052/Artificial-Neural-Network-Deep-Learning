```python
import tensorflow as tf
from tensorflow.keras.layers import Attention

# Query
query = tf.random.normal((1, 3, 128))

# Value
value = tf.random.normal((1, 5, 128))

# Attention layer
attention = Attention()

# Apply attention
context = attention([
    query,
    value
])

print("Query shape:", query.shape)
print("Value shape:", value.shape)
print("Context shape:", context.shape)
```

```python
import matplotlib.pyplot as plt

# Input words
words = ["I", "love", "machine", "learning"]

# Example attention weights
attention_weights = [0.05, 0.10, 0.20, 0.65]

# Create graph
plt.figure(figsize=(10, 5))

plt.bar(words, attention_weights)

plt.title("Attention Mechanism")
plt.xlabel("Input Words")
plt.ylabel("Attention Weight")

plt.ylim(0, 1)
plt.grid(axis="y")

plt.show()
```
